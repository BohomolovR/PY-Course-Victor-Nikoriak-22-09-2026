# 📦 Фінальний проєкт — Таємнича коробка Нової Пошти

Кожна група вирішує своє завдання і **лідер групи** відправляє результат.

```
         📦 BOX
           ↑
  ┌────────┼────────┬────────┐
  Група 1  Група 2  Група 3  Група 4
  Адреси   Вага    Логістика Безпека
```

> ⚠️ Якщо хоча б одна група помилиться — коробка не відкриється.

````python
# 🔒 SYSTEM CELL — запустити один раз

import requests
import csv
import io
import time

SYSTEM_READY = True
STUDENT_NAME = None

FINAL_FORM_URL = "https://docs.google.com/forms/d/e/1FAIpQLSe5oUb8rI8lz5ud_oCQDvj0_s6FUYHjFU-XitEgb94z0s3zdQ/formResponse"
FINAL_SHEET_CSV = "https://docs.google.com/spreadsheets/d/e/2PACX-1vQ16oP8S-HwlXJsWhXtXpeUdIdZHV1mBnn9ipYQe0Dz57W4BnvGxkBG11NRatXY49SVOiGKFnFYe0Aw/pub?gid=2126041300&single=true&output=csv"

def require_student():
    if STUDENT_NAME is None or str(STUDENT_NAME).strip() == "":
        print("⛔ Спочатку представтесь у клітинці 👤")
        return False
    return True

def submit_group_result(group, value):
    """Відправляє результат групи у Google Sheet."""
    if not require_student():
        return

    data = {
        "entry.575335391": STUDENT_NAME,
        "entry.1320852544": "FinalProject",
        "entry.705949827": "group_result",
        "entry.678121911": group,
        "entry.856775041": str(value)
    }

    try:
        requests.post(FINAL_FORM_URL, data=data)
        print(f"✅ Результат {group} відправлено: {value}")
    except Exception as e:
        print(f"❌ Помилка відправки: {e}")

def read_group_results():
    """Читає останні результати кожної групи зі Sheet."""
    try:
        response = requests.get(FINAL_SHEET_CSV, timeout=10)
        reader = csv.DictReader(io.StringIO(response.text))
        results = {}
        for row in reader:
            group = row.get("group", "").strip()
            result = row.get("result", "").strip()
            if group.startswith("GROUP"):
                results[group] = result  # перезаписує → залишається останній
        return results
    except Exception as e:
        print(f"❌ Помилка читання Sheet: {e}")
        return {}

print("🔧 System ready ✅")
````

## 🚀 Фінальна перевірка

> ⚠️ **Цю клітинку запускає тільки викладач** після того, як всі групи відправили результати.

````python
# 🚀 ФІНАЛЬНА ПЕРЕВІРКА — тільки викладач

import requests, csv, io, time

FINAL_SHEET_CSV = "https://docs.google.com/spreadsheets/d/e/2PACX-1vQ16oP8S-HwlXJsWhXtXpeUdIdZHV1mBnn9ipYQe0Dz57W4BnvGxkBG11NRatXY49SVOiGKFnFYe0Aw/pub?gid=2126041300&single=true&output=csv"

# ── Читаємо Sheet ────────────────────────────────────────
print("📡 Підключення до системи Нової Пошти...")
time.sleep(2)

try:
    response = requests.get(FINAL_SHEET_CSV, timeout=10)
    reader = csv.DictReader(io.StringIO(response.text))
    results = {}
    for row in reader:
        group = row.get("group", "").strip()
        result = row.get("result", "").strip()
        if group.startswith("GROUP"):
            results[group] = result
except Exception as e:
    print(f"❌ Не вдалося прочитати дані: {e}")
    results = {}

# ── Перевірка кожної групи ───────────────────────────────
print("\n" + "=" * 45)
print("📦 ПЕРЕВІРКА КОРОБКИ")
print("=" * 45)

all_ok = True

# Група 1
print("\n🔍 Перевірка Адресного відділу...")
time.sleep(2)
g1 = results.get("GROUP1", None)
if str(g1) == "100":
    print("✅ Група 1 — Адреса OK")
else:
    print(f"❌ Група 1 — Помилка адреси. Отримано: {g1}, очікувалось: 100")
    all_ok = False

# Група 2
print("\n🔍 Перевірка Відділу ваги...")
time.sleep(2)
g2 = results.get("GROUP2", None)
try:
    g2_val = int(g2)
    if g2_val == 25:
        print("✅ Група 2 — Вага OK")
    else:
        print(f"❌ Група 2 — Невірна вага. Отримано: {g2_val}, очікувалось: 25")
        all_ok = False
except (TypeError, ValueError):
    print(f"❌ Група 2 — Невірний тип. Отримано: {g2}")
    all_ok = False

# Група 3
print("\n🔍 Перевірка Логістики...")
time.sleep(2)
g3 = results.get("GROUP3", None)
try:
    g3_val = int(g3)
    if g3_val == 55:
        print("✅ Група 3 — Логістика OK")
    else:
        print(f"❌ Група 3 — Невірна ціна. Отримано: {g3_val}, очікувалось: 55")
        all_ok = False
except (TypeError, ValueError):
    print(f"❌ Група 3 — Помилка даних. Отримано: {g3}")
    all_ok = False

# Група 4
print("\n🔍 Перевірка Служби безпеки...")
time.sleep(3)
print("⚠️  Верифікація коду доступу...")
time.sleep(3)
g4 = results.get("GROUP4", None)
if g4 == "bcd":
    print("✅ Група 4 — Код безпеки OK")
else:
    print(f"❌ Група 4 — Невірний код. Отримано: {g4}, очікувалось: bcd")
    all_ok = False

# ── Фінал ────────────────────────────────────────────────
print("\n" + "=" * 45)
time.sleep(2)

if all_ok:
    print("🔓 КОРОБКУ ВІДКРИТО!")
    time.sleep(1)
    print("🎉 Усі відділи спрацювали правильно!")
    time.sleep(1)
    print()
    print("📦 Contents detected:")
    print("   Python Understanding Level ↑")
else:
    print("🔒 Коробка залишається закритою.")
    print("   Знайдіть помилку та спробуйте ще раз.")

print("=" * 45)
````
